## BAR PLOT

A bar plot in Matplotlib is a type of chart that represents data using rectangular bars. Each bar's height (or length, in horizontal bars) corresponds to the value it represents. Bar plots are useful for comparing quantities across different categories.

🔹 Definition in Matplotlib:
In Matplotlib, a bar plot is created using the plt.bar() or plt.barh() function from the matplotlib.pyplot module.

```python
Units_Sold1= [25, 40, 45, 50, 40, 25, 55]
Product= ['Product A', 'Product B', 'Product D', 'Product E', 'Product F', 'Product G', 'Product H']
```

## Histogram

In statistics, it is essential to be aware of the distribution of continuous variables prior to running any type of analysis. To display the distribution, we will use a histogram. Histograms display the frequency by the bin for a given array

In Matplotlib, a histogram is a type of plot that shows the distribution of a dataset. It divides the data into bins (intervals) and displays the frequency (count) of data points in each bin as a bar.

🔹 Definition:
A histogram is a graphical representation used to visualize the frequency distribution of a continuous variable. In Matplotlib, it's created using the plt.hist() function.

## Stacked Bar

A stacked plot in Matplotlib is a type of plot where multiple data series are stacked on top of each other. It is often used to show how individual components contribute to the total over time or categories.

Definition:
A stacked plot (like a stacked line or area chart) displays different groups of data on top of each other in layers, making it easy to see both individual and cumulative trends over time or categories.

```python
import pandas as pd
data={
    'Quarter': ['Q1','Q2','Q3','Q4'],
    'Product A': [10,20,30,40],
    'Product B':[20,25,35,45],
    'Product c':[30,30,40,50]
    }
df=pd.DataFrame(data)
```

```python
df
```

```python
df.set_index('Quarter',inplace=True)
```

```python
df
```

```python
df.plot(kind='bar',stacked=True,colormap='cividis',ylabel='sales',title='sales over quarters',figsize=(10,5));
```

## Pie Plot

In Matplotlib, a pie plot (or pie chart) is a circular statistical graphic used to show proportions of a whole. Each slice of the pie represents a category's contribution to the total.

```python
Units_Sold1= [25, 40, 45, 50]
Product= ['Product A', 'Product B', 'Product D', 'Product E']
```

```python
plt.pie(Units_Sold1,labels=Product,autopct="%1.2f%%",shadow=True,explode=(0,0,0,0.1),startangle=60,colors=['red','darkred','tomato','orange'])
plt.show()
```

## Scatter plot

* A scatter plot is a type of data visualization used to display the relationship between two numerical variables.
* Each point on the plot represents an observation, with its position determined by the values of the two variables.
* Scatter plots are useful for identifying trends, correlations, and patterns in data.

```python
import numpy as np
np.random.seed(42)
data={
    'Feature X':np.random.normal(loc=50,scale=10,size=100),
    'Feature Y':np.random.normal(loc=60,scale=15,size=100)
}
```

```python
df=pd.DataFrame(data)
df
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['Feature X'],df['Feature Y'],alpha=0.5,edgecolor='black',color='red')
plt.xlabel('Feature X')
plt.ylabel('Feature Y')
plt.title('scatter plot of features X and Y')
plt.grid(True)
plt.show()
```

```python
data={
    'X':np.random.randint(25,100,size=200),
    'Y':np.random.randint(10,150,size=200)
}
```

```python
df=pd.DataFrame(data)
df
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['X'],df['Y'],alpha=0.6,edgecolor='black')
plt.title('SCATTER PLOT OF X AND Y')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

## Box Plot

* A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile (Q1), median, third quartile (Q3), and “maximum”). 
* It can tell you about your outliers and what their values are. It can also tell you if your data is symmetrical how tightly your data is grouped, and if and how your data is skewed.
* The IQR (the range between Q1 and Q3) forms the box, showing the spread of the middle 50% of the data.
* A smaller IQR indicates less variability in the central data, while a larger IQR suggests more spread in the middle values.

```python
windspeed=[10, 15, 8, 20, 12, 0, 7, 14, 21, 18, 9, 6, 3, 11, 5, 13, 17, 19, 22, 16, 8, 10, 7, 5, 13]
```

```python
plt.boxplot(windspeed)
plt.title('Wind distribution')
plt.show()
```

```python
import pandas as pd
import numpy as np
np.random.seed(42)
data = {
    'Category A': np.random.normal(50, 10, 100),  # Normal distribution, mean=50, std=10
    'Category B': np.random.normal(60, 20, 100),  # Normal distribution, mean=60, std=20
    'Category C': np.random.normal(55, 5, 100),   # Normal distribution, mean=55, std=5
    'Category D': np.append(np.random.normal(45, 10, 95), [100, 105, 110, 115, 120])  # Skewed with outliers
}
```

## Violin Plot

* A violin plot is a hybrid of a box plot and a kernel density plot, which shows peaks in the data. 
* It is used to visualize the distribution of numerical data.
* Unlike a box plot that can only show summary statistics, violin plots depict summary statistics and the density of each variable.

```python
plt.violinplot(windspeed)
plt.show()
```

```python
plt.violinplot(data)
plt.show()
```

## Object-Oriented Approach Using Subplots

In Matplotlib, subplots allow you to create multiple plots (or axes) within a single figure. This is useful when you want to compare different visualizations side by side

* In the object-oriented approach, we create a figure object that acts as an empty canvas and then we add a set of axes, or subplots, to it
* Another advantage of using the object-oriented plotting approach is the ability to create multiple subplots on a single figure object.
* In some situations, we want to compare different views of data side by side. We can accomplish this in Matplotlib using subplots

```python
import numpy as np
a=np.linspace(0,10,20)
```

```python
a
```

```python
a2=a**2
```

```python
a2
```

```python
a3=a**3
```

```python
a3
```

```python
fig,axes=plt.subplots(nrows=1,ncols=2)
axes[0].plot(a,a2)
axes[0].set_title('A by A Squared')
axes[0].set_xlabel('Linearly spaced Numbers')
axes[0].set_ylabel('a2 value')

axes[1].plot(a,a3)
axes[1].set_title('A by A cubed')
axes[1].set_xlabel('Linearly spaced Numbers')
axes[1].set_ylabel('a3 value')
plt.tight_layout()
plt.show()
```

## Explanation:

* Scatter Plot: Shows how two continuous variables relate to each other.
* Line Plot: Illustrates trends over a continuous variable.
* Bar Plot: Compares average values of a numerical variable across categories.
* Box Plot: Visualizes the distribution and outliers of a numerical variable by category.
* Violin Plot: Combines box plot features with a density plot to show the distribution of data.
