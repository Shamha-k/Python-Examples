## Array Filtering

```python
arr=np.array([10,11,12,13,14,15])
x=[True,True,False,True,False,False]
newarr=arr[x]
newarr
```

```python
arr=np.arange(1,15)
filter_arrays=[]
for i in arr:
    if i%2==0:
        filter_arrays.append(True)
    else:
        filter_arrays.append(False)
newarr=arr[filter_arrays]
print(newarr)
```

```python
arr
```

```python
filter_arrays
```

### array filter directly

```python
arr=np.arange(1,15)
filter_array=arr%2!=0
newarr=arr[filter_array]
newarr
```
