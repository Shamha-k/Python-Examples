```python
marks=pd.Series([65,76,45,87])
marks
```

#### if we want a column

```python
df.loc[:,'age']
```

```python
In pandas, the colon means “select everything in this dimension.”
So in:


- The first position (before the comma) is for rows
- The second position (after the comma) is for columns
Thus:
- : → all rows
- 'age' → only the age column
So the line means:
“Select all rows, but only the age column.”



```

```python
df.loc[0:2,'name']
```

loc → label

iloc → index position

Each column here is a Series.
